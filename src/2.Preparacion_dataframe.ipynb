# 2. Preparación del dataframe final

## 2.1. Preparación inicial

```python
#Importar librerias
import pandas as pd
from utils import min_max_transformation
```

```python
#Cargar el dataframe crudo
df_pokemon_raw = pd.read_csv("./data/df_pokeapi_raw.csv", sep=",", index_col=0)
```

```python
#Hacemos una copia sobre la que trabajar
df_pokemon = df_pokemon_raw.copy()
```

## 2.2. Transformar variables height y weight a sistema internacional

```python
#Vamos a convertir height de decimetros (dm) a metros(m)
##Renombramos la columna
df_pokemon = df_pokemon.rename(columns={"height":"height_dm"})
##Generamos una nueva columna "height_m"
df_pokemon["height_m"] = df_pokemon["height_dm"]/10
#Vamos a convertir weight de hectograms (hg) a kilogramos (kg)
##Renombramos la columna
df_pokemon = df_pokemon.rename(columns={"weight":"weight_hg"})
##Generamos una nueva columna "weight_kg"
df_pokemon["weight_kg"] = df_pokemon["weight_hg"]/10
```

```python
#Vamos a quitar las columnas originales
df_pokemon.drop(columns=['height_dm', 'weight_hg'], inplace=True)
```

```python
df_pokemon
```

## 2.3. Obtenemos escalados min-max de variables de interés

### Variables derivadas de stats de combate

```python
#Vamos a calcular variables con escalado min-max para las variables stats
##Primero comprobamos rangos
print(min_max_transformation(df_pokemon,"stat_hp").describe(),"\n")
print(min_max_transformation(df_pokemon,"stat_attack").describe(),"\n")
print(min_max_transformation(df_pokemon,"stat_special-attack").describe(),"\n")
print(min_max_transformation(df_pokemon,"stat_defense").describe(),"\n")
print(min_max_transformation(df_pokemon,"stat_special-defense").describe(),"\n")
print(min_max_transformation(df_pokemon,"stat_speed").describe(),"\n")
```

```python
#Guardamos los valores minmax en el dataframe
df_pokemon["minmax_stat_hp"] = min_max_transformation(df_pokemon,"stat_hp")
df_pokemon["minmax_stat_attack"] = min_max_transformation(df_pokemon,"stat_attack")
df_pokemon["minmax_stat_special-attack"] = min_max_transformation(df_pokemon,"stat_special-attack")
df_pokemon["minmax_stat_defense"] = min_max_transformation(df_pokemon,"stat_defense")
df_pokemon["minmax_stat_special-defense"] = min_max_transformation(df_pokemon,"stat_special-defense")
df_pokemon["minmax_stat_speed"] = min_max_transformation(df_pokemon,"stat_speed")
```

```python
df_pokemon
```

### Calcular avg. minmax combat score

```python
#Calculamos la media de los minmax de los stats de combate
minmax_stat_cols = ["minmax_stat_hp", "minmax_stat_attack", "minmax_stat_special-attack", "minmax_stat_defense", "minmax_stat_special-defense", "minmax_stat_speed"]
(df_pokemon[minmax_stat_cols].sum(axis=1) / len(minmax_stat_cols)).describe()
```

```python
#Guardamos el score
df_pokemon["avg_minmax_stats_combat"] = df_pokemon[minmax_stat_cols].sum(axis=1) / len(minmax_stat_cols)
```

### Generamos columnas buleanas para los types

```python
#Tipo principal
(df_pokemon.type_1).unique()
```

```python
#Tipo secundario
(df_pokemon.type_2).unique()
```

```python
tipos_en_ambos = list(set(df_pokemon['type_1']) & set(df_pokemon['type_2']))
print(len(tipos_en_ambos))
tipos_en_ambos
```

Como podemos ver tenemos un total de 18 tipos posibles pero a veces pueden ser el tipo principal (type_1) o secundario(type_2). Así que para facilitar análisis posteriores vamos a generar columnas buleanas que luego podamos usar para gráficos o análisis.

```python
for tipe in tipos_en_ambos:
   df_pokemon["type_"+tipe] = ((df_pokemon['type_1']==tipe) | (df_pokemon['type_2']==tipe))
```
